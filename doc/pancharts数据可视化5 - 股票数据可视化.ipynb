## sk_vm k线图的绘制


* 股票数据框的规则，索引为日期字符串，前四列为开盘价，收盘价，最低价，最高价
* 均线的添加


```python
from pancharts import sk_vm
import pandas as pd

# 创建股票数据
data = pd.DataFrame({
    'open': [34, 35, 38, 33, 36, 40],
    'close': [36, 38, 36, 35, 40, 42],
    'low': [30, 32, 35, 30, 34, 38],
    'high': [38, 40, 42, 38, 42, 45]
}, index=['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05', '2024-01-08'])

# 创建K线图实例
chart = sk_vm(data)

# 绘制K线图，添加5日和20日均线
result = chart.kline(2)
result.render_notebook()
```

## 通过akshare获取股票数据

```python
import akshare as ak

#stock_data = ak.stock_zh_a_hist(symbol="000002", period="daily", start_date="20220301", end_date='20240528', adjust="")
stock_data = ak.stock_zh_a_daily(symbol="sz000002", start_date="20240403", end_date="20250404", adjust="qfq")
#stock_data = ak.stock_zh_a_hist_tx(symbol="sz000002", start_date="20200101", end_date="20231027", adjust="")
```

```python
stock_data.close.values[0]
```

```python
stock_data0=stock_data.set_index("date")[["open","close","low","high","volume","amount"]]
stock_data0.head()
```

```python
# 创建K线图实例
chart = sk_vm(stock_data0)

# 绘制K线图，添加5日、20日、60日均线
result = chart.kline([5,20,60])
result.render_notebook()
```
