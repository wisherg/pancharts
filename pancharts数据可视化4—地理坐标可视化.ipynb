## k_vm的geo地理坐标可视化

* 行索引需要是地点名称，前两列必须是经度和纬度

* 其他列是特征值，默认映射第三列的值为color，修改映射与create_visual_map类似，本质上图形函数就是调用它实现的

* geo地点不受母图限制，能提供更灵活的可视化

```python
print(gk_vm.__doc__)
```

```python
#把下面这个数据转化为数据框，name为索引,后面value分为四列，分别为经度、纬度、人口、收入
z=[
        {"name": "北京", "value": [116.46, 39.92, 800,90,"市"]},
        {"name": "上海", "value": [121.48, 31.22, 900,70,"市"]},
        {"name": "广州", "value": [113.23, 23.16, 700,50,"省"]},
        {"name": "深圳", "value": [114.07, 22.62, 600,40,"省"]},
        {"name": "成都", "value": [104.06, 30.67, 500,20,"省"]}
      ]
    
z1={'经度':[],'纬度':[],'人口':[],'收入':[],'类型':[]}
z1_index=[]
for i in z:
  z1['经度'].append(i['value'][0])
  z1['纬度'].append(i['value'][1])
  z1['人口'].append(i['value'][2])
  z1['收入'].append(i['value'][3])
  z1['类型'].append(i['value'][4])
  z1_index.append(i['name'])

geo_df=pd.DataFrame(z1,index=z1_index)
geo_df
```

```python
chinamap=gk_vm(geo_df,"china")
p1=chinamap.scatter()
p1.render_notebook()
```

```python
p1=chinamap.escatter()
p1.render_notebook()
```

## 基于高德地图的地理坐标可视化

* 暂时只能在独立页面上显示，不能在notebook中直接显示

```python
print(gk_vm_amap.__doc__)
```

```python
from pancharts import gk_vm_amap
import pandas as pd

# 创建地理数据
data = pd.DataFrame({
    'lng': [116.4074, 121.4737, 104.0668],
    'lat': [39.9042, 31.2304, 30.5728],
    'value': [100, 200, 150]
}, index=['北京', '上海', '成都'])

# 创建高德地图可视化实例（以"北京市"为中心）
amap_chart = gk_vm_amap(data)

# 绘制散点图
chart = amap_chart.heatmap()
chart.render()
```

```python
data
```

### 重要配置

* 官方内置样式：normal(标准)、light(浅灰)、dark(暗黑)

* // 1. 中心点：默认会计算经纬度值的中心点
  // 可以自定义其他地址名称，例如："上海"、"广州"等

  // 2. 初始缩放级别（2-18，数值越大越近）
  zoom: 10, 

```python
config={"init":{"width":"100%","height":"800px"},"amap":{"mapStyle": 'amap://styles/dark',"zoom":7}}
chart = amap_chart.scatter(config=config)
chart.render()
```
